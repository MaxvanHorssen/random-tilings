```python
import numpy as np
import matplotlib.pyplot as plt
import RandomTilings as RT
import RandomTilingsTMP as RT0
```

```python
def diff(theta,phi):
    return np.abs(np.sin((theta-phi)/2))

n = 350
tau = 10

alpha = 0.
beta = np.pi/25
gamma = 4*np.pi/5
delta = 6*np.pi/5
epsilon = 8*np.pi/5

r  = 2

w = np.array([[1+1j,diff(epsilon,gamma),diff(gamma,beta),diff(delta,gamma),diff(gamma,beta),diff(beta,gamma)],
[diff(gamma,beta),diff(beta,epsilon),1+r*1j,diff(beta,delta),1+1j,1+r*1j],
[diff(delta,gamma),diff(epsilon,delta),diff(delta,beta),1+1j,diff(delta,alpha),diff(alpha,delta)],
[diff(gamma,alpha),diff(alpha,epsilon),diff(beta,alpha),diff(alpha,delta),1+r*1j,1+1j]])

A = RT0.Aztec(n,w)
A.shuffle()
A.plot(edge=0,dpi=150,coloring='(255,33,33),(31,31,255),(32,255,32),(255,255,32)')
```

```python
def diff(theta,phi):
    return np.abs(np.sin((theta-phi)/2))
```

```python
n = 120*2
tau = 10

alpha = 0.
beta = np.pi/25
gamma = 4*np.pi/5
delta = 6*np.pi/5
epsilon = 8*np.pi/5


w = np.array([[np.exp(-tau),diff(epsilon,gamma),diff(gamma,beta),diff(delta,gamma),diff(gamma,beta),diff(beta,gamma)],
[diff(gamma,beta),diff(beta,epsilon),np.exp(-4*tau),diff(beta,delta),np.exp(-tau),np.exp(-4*tau)],
[diff(delta,gamma),diff(epsilon,delta),diff(delta,beta),np.exp(-tau),diff(delta,alpha),diff(alpha,delta)],
[diff(gamma,alpha),diff(alpha,epsilon),diff(beta,alpha),diff(alpha,delta),np.exp(-4*tau),np.exp(-tau)]])

A = RT.Aztec(n,w,hard_drive_mode=1)
A.shuffle()
A.plot(edge=0,dpi=150,rotated=False)
```
