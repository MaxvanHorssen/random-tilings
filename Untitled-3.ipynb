# Normal Routine

```python
@jit("(float64, float64, float64, float64)")
def red_cases(w,x,y,z):
    if w+x+y+z == 0:
        return 1/2**0.5,1/2**0.5,1/2**0.5,1/2**0.5
    elif z+x==0:
        return 1/(w+y),y,1/(w+y),w
    elif w+x==0:
        return z,y,1/(z+y),1/(z+y)
    elif w+y==0:
        return z,1/(z+x),x,1/(z+x)
    elif z+y==0:
        return 1/(x+w),1/(x+w),x,w
    else:
        tmp = z*w+x*y
        return z/tmp, y/tmp, x/tmp, w/tmp
            


@jit(["(Array(float64, 2, 'C', False, aligned=True),)","(Array(float64, 2, 'A', False, aligned=True),)"])
def reduction(W):
    N = W.shape[0]
    W_red = np.zeros((N,N))

    for m in range(N//2):
        for n in range(N//2):
            w,x,y,z = W[2*m,2*n],W[2*m,2*n+1],W[2*m+1,2*n],W[2*m+1,2*n+1]
            p,q,r,s = red_cases(w,x,y,z)
            W_red[2*m,2*n]     = p
            W_red[2*m,2*n+1]   = q
            W_red[2*m+1,2*n]   = r
            W_red[2*m+1,2*n+1] = s

    for m in range(N//2):
        for n in range(N//2):
            w,x,y,z = W[2*m,2*n],W[2*m,2*n+1],W[2*m+1,2*n],W[2*m+1,2*n+1]
            if w+x == 0:
                W_red[2*m-1,2*n],W_red[2*m-1,2*n+1]=0,0
            if x+z == 0:
                W_red[2*m,2*n+2],W_red[2*m+1,2*n+2]=0,0
            if y+z == 0:
                W_red[2*m+2,2*n],W_red[2*m+2,2*n+1]=0,0
            if w+y == 0:
                W_red[2*m,2*n-1],W_red[2*m+1,2*n-1]=0,0
        
    return W_red[1:-1,1:-1]



@njit("(Array(float64, 2, 'C', False, aligned=True),)")
def reduce_weight(W):
    K = W.shape[0] // 2
    W = W.copy()
    C = List()
    C.append(np.ascontiguousarray(W))

    for _ in range(2, K + 1):
        R = reduction(C[-1])
        C.append(np.ascontiguousarray(R))

    return C[::-1]



@jit( "(Array(int8, 2, 'C', False, aligned=True), int64, int64, int64)")
def no_neighbor(M,k,m,n):
    N = M.shape[0]
    A = N//2-1-k  
    
    if M[A+2*m,A+2*n]+M[A+2*m+1,A+2*n]+M[A+2*m,A+2*n+1]+M[A+2*m+1,A+2*n+1]>0:
        return False

    if n==0:
        tmp_w = True
        tmp_e = (M[A+2*m,A+2*(n+1)]+M[A+2*m+1,A+2*(n+1)])==0
    elif n==k:
        tmp_w = (M[A+2*m,A+2*n-1]+M[A+2*m+1,A+2*n-1])==0
        tmp_e = True
    else:
        tmp_w = (M[A+2*m,A+2*n-1]+M[A+2*m+1,A+2*n-1])==0
        tmp_e = (M[A+2*m,A+2*(n+1)]+M[A+2*m+1,A+2*(n+1)])==0

    if m==0:
        tmp_n = True
        tmp_s = (M[A+2*(m+1),A+2*n]+M[A+2*(m+1),A+2*n+1])==0
    elif m==k:
        tmp_n = (M[A+2*m-1,A+2*n]+M[A+2*m-1,A+2*n+1])==0
        tmp_s = True
    else:
        tmp_n = (M[A+2*m-1,A+2*n]+M[A+2*m-1,A+2*n+1])==0
        tmp_s = (M[A+2*(m+1),A+2*n]+M[A+2*(m+1),A+2*n+1])==0

    return tmp_s and tmp_e and tmp_w and tmp_n


@jit("(ListType(Array(float64, 2, 'C')),)")
def shuffling(C):
    N = 2*len(C)
    M = np.zeros((N,N),dtype=np.int8)
    #print(N)

    # Initiate
    A = N//2-1
    w,x,y,z = C[0][0,0],C[0][0,1],C[0][1,0],C[0][1,1]

    tmp = (w*z+x*y)


    prob = (w*z)/tmp
    if np.random.random()<prob:
        M[A,A],M[A+1,A+1]= 1,1
        M[A,A+1],M[A+1,A]= 0,0
    else:
        M[A,A],M[A+1,A+1]= 0,0
        M[A,A+1],M[A+1,A]= 1,1

    for k in range(1,N//2):
        A = N//2-1-k   
        

        # destruction + flip
        for m in range(k+1):
            for n in range(k+1):
                tmp = M[A+2*m,A+2*n]+M[A+2*m+1,A+2*n]+M[A+2*m,A+2*n+1]+M[A+2*m+1,A+2*n+1]
                if tmp>1:
                    M[A+2*m,A+2*n],M[A+2*m+1,A+2*n],M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n+1]=0,0,0,0
                elif tmp == 1:
                    M[A+2*m,A+2*n],M[A+2*m+1,A+2*n+1]=M[A+2*m+1,A+2*n+1],M[A+2*m,A+2*n]
                    M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n]=M[A+2*m+1,A+2*n],M[A+2*m,A+2*n+1]

        for m in range(k+1):
            for n in range(k+1):
                if no_neighbor(M,k,m,n):
                    w,x,y,z = C[k][2*m,2*n],C[k][2*m,2*n+1],C[k][2*m+1,2*n],C[k][2*m+1,2*n+1]
                    prob = (w*z)/(w*z+x*y)
                    if np.random.random()<prob:
                        M[A+2*m,A+2*n],M[A+2*m+1,A+2*n+1]= 1,1
                        M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n]= 0,0
                    else:
                        M[A+2*m,A+2*n],M[A+2*m+1,A+2*n+1]= 0,0
                        M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n]= 1,1
    return M



def draw_aztec(n,w,edge=0,paths=False,dots=False,rotated=True,coloring='standard',dpi=100,show_figure=True):
    w = w.astype(float)
    edge = str(edge)
    W = weight_aztec.weight_aztec(n,w)
    C = reduce_weight(W)
    M = shuffling(C).astype(int)
    fig = draw_dominos.draw_dominos(M,False,edge,paths,dots,rotated,coloring,False,dpi,show_figure)
    return fig
```

# Memory Efficient

```python
@jit("(Array(float64, 2, 'C', False, aligned=True),)")
def reduce_weight1(W):
    K = W.shape[0] // 2
    C = W.copy()
    for k in range(K-1):
        for m in range(K-k):
            for n in range(K-k):
                d = C[k+2*m,k+2*n]*C[k+2*m+1,k+2*n+1]+C[k+2*m+1,k+2*n]*C[k+2*m,k+2*n+1]
                w,x,y,z = C[k+2*m,k+2*n],C[k+2*m,k+2*n+1],C[k+2*m+1,k+2*n],C[k+2*m+1,k+2*n+1]
                C[k+2*m,k+2*n]     = z/d
                C[k+2*m,k+2*n+1]   = y/d
                C[k+2*m+1,k+2*n]   = x/d
                C[k+2*m+1,k+2*n+1] = w/d
    return C

@jit("(Array(float64, 2, 'C', False, aligned=True), int64)")
def shuffling1(C,N):
    M = np.zeros((N,N),dtype=np.int8)
    #print(N)

    # Initiate
    A = N//2-1
    w,x,y,z = C[A,A],C[A,A+1],C[A+1,A],C[A+1,A+1]

    tmp = (w*z+x*y)


    prob = (w*z)/tmp
    if np.random.random()<prob:
        M[A,A],M[A+1,A+1]= 1,1
        M[A,A+1],M[A+1,A]= 0,0
    else:
        M[A,A],M[A+1,A+1]= 0,0
        M[A,A+1],M[A+1,A]= 1,1

    for k in range(1,N//2):
        A = N//2-1-k   

        # destruction + flip (+ reconstruct)
        for m in range(k+1):
            for n in range(k+1):
                # reconstruct
                w,x,y,z = C[A+2*m,A+2*n],C[A+2*m+1,A+2*n],C[A+2*m,A+2*n+1],C[A+2*m+1,A+2*n+1]
                d = w*z+x*y
                C[A+2*m,A+2*n],C[A+2*m+1,A+2*n+1] = C[A+2*m+1,A+2*n+1]/d, C[A+2*m,A+2*n]/d
                C[A+2*m,A+2*n+1],C[A+2*m+1,A+2*n] = C[A+2*m+1,A+2*n]/d, C[A+2*m,A+2*n+1]/d

                # destruction + flip
                tmp = M[A+2*m,A+2*n]+M[A+2*m+1,A+2*n]+M[A+2*m,A+2*n+1]+M[A+2*m+1,A+2*n+1]
                if tmp>1:
                    M[A+2*m,A+2*n],M[A+2*m+1,A+2*n],M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n+1]=0,0,0,0
                elif tmp == 1:
                    M[A+2*m,A+2*n],M[A+2*m+1,A+2*n+1]=M[A+2*m+1,A+2*n+1],M[A+2*m,A+2*n]
                    M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n]=M[A+2*m+1,A+2*n],M[A+2*m,A+2*n+1]

        for m in range(k+1):
            for n in range(k+1):
                if no_neighbor(M,k,m,n):
                    w,x,y,z = C[A+2*m,A+2*n],C[A+2*m,A+2*n+1],C[A+2*m+1,A+2*n],C[A+2*m+1,A+2*n+1]
                    prob = (w*z)/(w*z+x*y)
                    if np.random.random()<prob:
                        M[A+2*m,A+2*n],M[A+2*m+1,A+2*n+1]= 1,1
                        M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n]= 0,0
                    else:
                        M[A+2*m,A+2*n],M[A+2*m+1,A+2*n+1]= 0,0
                        M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n]= 1,1
    return M

def draw_aztec1(n,w,edge=0,paths=False,dots=False,rotated=True,coloring='standard',dpi=100,show_figure=True):
    w = w.astype(float)
    edge = str(edge)
    W = weight_aztec.weight_aztec(n,w)
    C = reduce_weight1(W)
    M = shuffling1(C,2*n).astype(int)
    fig = draw_dominos.draw_dominos(M,False,edge,paths,dots,rotated,coloring,False,dpi,show_figure)
    return fig
```

# Compare Timings

```python
from RandomTilings.algorithm_reduction_weight import algorithm_reduction_weight
from RandomTilings.shuffling import shuffling as shuffling_christophe

from tqdm import trange
```

```python
# All test without plotting!!!

def aztec_christophe(n,w):
    w = w.astype(float)
    W = weight_aztec.weight_aztec(n,w)
    C = algorithm_reduction_weight(W)
    M = shuffling_christophe(C).astype(int)

def aztec_fast(n,w):    
    w = w.astype(float)
    W = weight_aztec.weight_aztec(n,w)
    C = reduce_weight(W)
    M = shuffling(C).astype(int)

def aztec_memory(n,w):
    w = w.astype(float)
    W = weight_aztec.weight_aztec(n,w)
    C = reduce_weight1(W)
    M = shuffling1(C,2*n).astype(int)
```

```python
w = np.array([[1]])

t_christ = []
t_fast   = []
t_memory = []

N = trange(1,101)
reps = 20
for n in N:
    t1 = time()
    for _ in range(reps):
        aztec_christophe(n,w)
    t_christ += [(time()-t1)/reps]

    t1 = time()
    for _ in range(reps):
        aztec_fast(n,w)
    t_fast += [(time()-t1)/reps]

    t1 = time()
    for _ in range(reps):
        aztec_memory(n,w)
    t_memory += [(time()-t1)/reps]
```

```python
plt.plot(range(1,len(t_christ)+1),t_christ,label='Christophe')
plt.plot(range(1,len(t_fast)+1),t_fast,label = 'Fast')
plt.plot(range(1,len(t_memory)+1),t_memory, label = 'Memory')
plt.legend()
plt.show()
```
