# Log partition function

```python
n = 2
a,b,c = 1,2,3
H = RT.Hexagon(n,a=a,b=b,c=c)
logPartitionFunction = H.get_log_partition_function()
H.close()
print("Do valid tilings exist?", logPartitionFunction.imag==0)

from numpy import exp
partitionFunction = exp(logPartitionFunction.real)
print(partitionFunction)

# MacMahon's formula
A,B,C = a*n,b*n,c*n
exactPartitionFunction = 1
for i in range(1,A+1):
    for j in range(1,B+1):
        for k in range(1,C+1):
            exactPartitionFunction *= (i+j+k-1)/(i+j+k-2)
print(exactPartitionFunction)
```
