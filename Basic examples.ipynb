```python
import RandomTilings as RT
```

# Aztec diamond

```python
A = RT.Aztec(10)
A.shuffle()
A.plot()
A.plot(edge_width=1,orientation='square')
```

```python
a = 0.5
b = 1
w = [[b,b,a,a],[b,b,a,a],[a,a,b,b],[a,a,b,b]]
A = RT.Aztec(100,w)
A.shuffle()
A.plot(color_scheme='aztec gray')
```

```python
gap = [[9,-2,2],[15,3,5]]
A = RT.Aztec(10,gap=gap)
A.shuffle()
A.plot(dot_width=1,edge_width=1,gap_width=2,orientation='square',path_width=1)
```

```python
gap = [[2*150-1,80,150]]
A = RT.Aztec(300,gap=gap)
A.shuffle()
A.plot()
```

# Hexagon

```python
H = RT.Hexagon(3,a=1,b=2,c=3)
H.shuffle()
H.plot()
H.plot(edge_width=1,shape='skewed')
```

```python
a1 = 0.3
a2 = 0.3
w = [[1,1,1],[a2/a1,a1/a2,1],[1,1,1],[1/a2,a2,1],[1,1,1],[a1,1/a1,1]]
H = RT.Hexagon(100,w)
H.shuffle()
H.plot(color_scheme=[(5,50,100),(120,180,200),(135,60,35)])
# This custom color scheme is equilvalent with color_scheme='alternative'
```

```python
gap = [[5,3.5,5.5]]
H = RT.Hexagon(5,gap=gap)
H.shuffle()
H.plot(dot_width=1,edge_width=1,gap_width=2,path_width=1,shape='skewed')
```

```python
n = 1
a,b,c = 120,120,80
gap = [[130,100.5,140.5]]
H = RT.Hexagon(n,a=a,b=b,c=c,gap=gap)
H.shuffle()
H.plot()
```
