# Aztec diamond

```python
A = RT.Aztec(10)
A.shuffle()
A.plot()
A.plot(border_size=1,boundary_size=2,orientation='square')
```

```python
a = 0.5
b = 1
w = [[b,b,a,a],[b,b,a,a],[a,a,b,b],[a,a,b,b]]
A = RT.Aztec(100,w)
A.shuffle()
A.plot(color_scheme='aztec gray')
```

```python
gap = [[9,-2,2],[15,3,5]]
A = RT.Aztec(10,gap=gap)
A.shuffle()
A.plot(border_size=1,orientation='square',color_scheme='white')
A.plot_paths(path_size=2,dot_size=20,gap_size=3)
```

```python
gap = [[2*150-1,80,150]]
A = RT.Aztec(300,gap=gap)
A.shuffle()
A.plot()
```

```python
a = 0.5
n = 5
A = RT.Aztec(n,[[a,1],[1,a]])
logPartitionFunction = A.get_log_partition_function()
print("Do valid tilings exist?", logPartitionFunction.imag==0)

from numpy import exp
partitionFunction = exp(logPartitionFunction.real)
print(partitionFunction)

# Exact value according to Elkies, Kuperberg, Larsen, and Propp (1992)
exactPartitionFunction = (1+a**2)**(n*(n+1)/2)
print(exactPartitionFunction)
```

# Hexagon

```python
H = RT.Hexagon(3,a=1,b=2,c=3)
H.shuffle()
H.plot()
H.plot(border_size=1,boundary_size=2,shape='skewed')
```

```python
a1 = 0.3
a2 = 0.3
w = [[1,1,1],[a2/a1,a1/a2,1],[1,1,1],[1/a2,a2,1],[1,1,1],[a1,1/a1,1]]
H = RT.Hexagon(100,w)
H.shuffle()
H.plot(color_scheme=[(5,50,100),(120,180,200),(135,60,35)])
# This custom color scheme is equilvalent with color_scheme='alternative'
```

```python
gap = [[5,3.5,5.5]]
H = RT.Hexagon(5,gap=gap)
H.shuffle()
H.plot(border_size=1,shape='skewed',color_scheme='white')
H.plot_paths(path_size=2,dot_size=20,gap_size=3)
```

```python
n = 1
a,b,c = 120,120,80
gap = [[130,100.5,140.5]]
H = RT.Hexagon(n,a=a,b=b,c=c,gap=gap)
H.shuffle()
H.plot()
```

```python
n = 2
a,b,c = 1,2,3
H = RT.Hexagon(n,a=a,b=b,c=c)
logPartitionFunction = H.get_log_partition_function()
print("Do valid tilings exist?", logPartitionFunction.imag==0)

from numpy import exp
partitionFunction = exp(logPartitionFunction.real)
print(partitionFunction)

# MacMahon's formula
A,B,C = a*n,b*n,c*n
exactPartitionFunction = 1
for i in range(1,A+1):
    for j in range(1,B+1):
        for k in range(1,C+1):
            exactPartitionFunction *= (i+j+k-1)/(i+j+k-2)
print(exactPartitionFunction)
```
