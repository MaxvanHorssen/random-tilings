```python

def edge_inclusion_prob(C):
    N = 2*len(C)
    P = np.zeros((N,N))

    A = N//2 -1 
    # Initiation
    w,x,y,z = C[0][0,0],C[0][0,1],C[0][1,0],C[0][1,1]
    tmp     = 1/(w*z+x*y)

    P[A,A]     = tmp*w*z
    P[A,A+1]   = tmp*x*y
    P[A+1,A]   = tmp*x*y
    P[A+1,A+1] = tmp*w*z



    for k in range(1,N//2):
        # A <- Index Anchor
        A = N//2-1-k     
                        
        # update P
        for m in range(k+1):
            for n in range(k+1):
                P[A+2*m,A+2*n],P[A+2*m+1,A+2*n+1] = P[A+2*m+1,A+2*n+1],P[A+2*m,A+2*n]
                P[A+2*m,A+2*n+1],P[A+2*m+1,A+2*n] = P[A+2*m+1,A+2*n],P[A+2*m,A+2*n+1]

                p,q,r,s = P[A+2*m,A+2*n],P[A+2*m,A+2*n+1],P[A+2*m+1,A+2*n],P[A+2*m+1,A+2*n+1]
                w,x,y,z = C[k][2*m,2*n],C[k][2*m,2*n+1],C[k][2*m+1,2*n],C[k][2*m+1,2*n+1]
                tmp     = (1-p-q-r-s)/(w*z+x*y)

                P[A+2*m+1,A+2*n+1] = s+tmp*w*z
                P[A+2*m+1,A+2*n]   = r+tmp*x*y
                P[A+2*m,A+2*n+1]   = q+tmp*x*y
                P[A+2*m,A+2*n]     = p+tmp*w*z

    return P
```

```python
from numba import jit,njit
from numba.typed import List

@jit("(float64, float64, float64, float64)")
def red_cases(w,x,y,z):
    if w+x+y+z == 0:
        return 1/2**0.5,1/2**0.5,1/2**0.5,1/2**0.5
    elif z+x==0:
        return 1/(w+y),y,1/(w+y),w
    elif w+x==0:
        return z,y,1/(z+y),1/(z+y)
    elif w+y==0:
        return z,1/(z+x),x,1/(z+x)
    elif z+y==0:
        return 1/(x+w),1/(x+w),x,w
    else:
        tmp = z*w+x*y
        return z/tmp, y/tmp, x/tmp, w/tmp
            

@jit(["(Array(float64, 2, 'C', False, aligned=True),)","(Array(float64, 2, 'A', False, aligned=True),)"])
def reduction(W):
    N = W.shape[0]
    W_red = np.zeros((N,N))

    for m in range(N//2):
        for n in range(N//2):
            w,x,y,z = W[2*m,2*n],W[2*m,2*n+1],W[2*m+1,2*n],W[2*m+1,2*n+1]
            p,q,r,s = red_cases(w,x,y,z)
            W_red[2*m,2*n]     = p
            W_red[2*m,2*n+1]   = q
            W_red[2*m+1,2*n]   = r
            W_red[2*m+1,2*n+1] = s

    for m in range(N//2):
        for n in range(N//2):
            w,x,y,z = W[2*m,2*n],W[2*m,2*n+1],W[2*m+1,2*n],W[2*m+1,2*n+1]
            if w+x == 0:
                W_red[2*m-1,2*n],W_red[2*m-1,2*n+1]=0,0
            if x+z == 0:
                W_red[2*m,2*n+2],W_red[2*m+1,2*n+2]=0,0
            if y+z == 0:
                W_red[2*m+2,2*n],W_red[2*m+2,2*n+1]=0,0
            if w+y == 0:
                W_red[2*m,2*n-1],W_red[2*m+1,2*n-1]=0,0
        
                
    return W_red[1:-1,1:-1]



@njit("(Array(float64, 2, 'C', False, aligned=True),)")
def reduce_weight(W):
    n = W.shape[0] // 2
    C = List()
    C.append(np.ascontiguousarray(W))

    for _ in range(2, n + 1):
        R = reduction(C[-1])
        C.append(np.ascontiguousarray(R))

    return C[::-1]


@jit( "(Array(int8, 2, 'C', False, aligned=True), int64, int64, int64)")
def no_neighbor(M,k,m,n):
    N = M.shape[0]
    A = N//2-1-k  
    
    if M[A+2*m,A+2*n]+M[A+2*m+1,A+2*n]+M[A+2*m,A+2*n+1]+M[A+2*m+1,A+2*n+1]>0:
        return False

    if n==0:
        tmp_w = True
        tmp_e = (M[A+2*m,A+2*(n+1)]+M[A+2*m+1,A+2*(n+1)])==0
    elif n==k:
        tmp_w = (M[A+2*m,A+2*n-1]+M[A+2*m+1,A+2*n-1])==0
        tmp_e = True
    else:
        tmp_w = (M[A+2*m,A+2*n-1]+M[A+2*m+1,A+2*n-1])==0
        tmp_e = (M[A+2*m,A+2*(n+1)]+M[A+2*m+1,A+2*(n+1)])==0

    if m==0:
        tmp_n = True
        tmp_s = (M[A+2*(m+1),A+2*n]+M[A+2*(m+1),A+2*n+1])==0
    elif m==k:
        tmp_n = (M[A+2*m-1,A+2*n]+M[A+2*m-1,A+2*n+1])==0
        tmp_s = True
    else:
        tmp_n = (M[A+2*m-1,A+2*n]+M[A+2*m-1,A+2*n+1])==0
        tmp_s = (M[A+2*(m+1),A+2*n]+M[A+2*(m+1),A+2*n+1])==0

    return tmp_s and tmp_e and tmp_w and tmp_n

@jit("(ListType(Array(float64, 2, 'C')),)")
def shuffling(C):
    N = 2*len(C)
    M = np.zeros((N,N),dtype=np.int8)
    #print(N)

    # Initiate
    A = N//2-1
    w,x,y,z = C[0][0,0],C[0][0,1],C[0][1,0],C[0][1,1]

    tmp = (w*z+x*y)


    prob = (w*z)/tmp
    if np.random.random()<prob:
        M[A,A],M[A+1,A+1]= 1,1
        M[A,A+1],M[A+1,A]= 0,0
    else:
        M[A,A],M[A+1,A+1]= 0,0
        M[A,A+1],M[A+1,A]= 1,1

    for k in range(1,N//2):
        A = N//2-1-k   
        

        # destruction + flip
        for m in range(k+1):
            for n in range(k+1):
                tmp = M[A+2*m,A+2*n]+M[A+2*m+1,A+2*n]+M[A+2*m,A+2*n+1]+M[A+2*m+1,A+2*n+1]
                if tmp>1:
                    M[A+2*m,A+2*n],M[A+2*m+1,A+2*n],M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n+1]=0,0,0,0
                elif tmp == 1:
                    M[A+2*m,A+2*n],M[A+2*m+1,A+2*n+1]=M[A+2*m+1,A+2*n+1],M[A+2*m,A+2*n]
                    M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n]=M[A+2*m+1,A+2*n],M[A+2*m,A+2*n+1]

        for m in range(k+1):
            for n in range(k+1):
                if no_neighbor(M,k,m,n):
                    w,x,y,z = C[k][2*m,2*n],C[k][2*m,2*n+1],C[k][2*m+1,2*n],C[k][2*m+1,2*n+1]
                    prob = (w*z)/(w*z+x*y)
                    if np.random.random()<prob:
                        M[A+2*m,A+2*n],M[A+2*m+1,A+2*n+1]= 1,1
                        M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n]= 0,0
                    else:
                        M[A+2*m,A+2*n],M[A+2*m+1,A+2*n+1]= 0,0
                        M[A+2*m,A+2*n+1],M[A+2*m+1,A+2*n]= 1,1



    return M







def draw_aztec1(n,w,edge=0,paths=False,dots=False,rotated=True,coloring='standard',dpi=100,show_figure=True):
    w = w.astype(float)
    edge = str(edge)
    W = weight_aztec.weight_aztec(n,w)
    C = reduce_weight(W)
    M = shuffling(C).astype(int)
    fig = draw_dominos.draw_dominos(M,False,edge,paths,dots,rotated,coloring,False,dpi,show_figure)
    return fig

def draw_hexagon1(n,w,a=1,b=1,c=1,skewed_grid=False,edge=0,paths=False,dots=False,coloring='standard',dpi=200,show_figure=True):
    w = w.astype(float)
    edge = str(edge)
    W = weight_hexagon.weight_hexagon(n,w,a,b,c)
    C = reduce_weight(W)
    M = shuffling(C).astype(int)
    fig = draw_lozenges.draw_lozenges(n,M,False,a,b,c,skewed_grid,edge,paths,dots,coloring,False,dpi,show_figure)
    return fig

def draw_aztec_gap1(n,w,gap,edge=0,paths=False,dots=False,rotated=True,coloring='standard',show_gap=False,dpi=200,show_figure=True):
    N = 2*n
    end = N

    w = w.astype(float)
    W = weight_aztec.weight_aztec(n,w)

    size_gap = gap.shape[0]
    for i1 in range(size_gap):
        X = gap[i1][0]
        for Y in range(gap[i1][1],gap[i1][2]+1):
            if X%2 == 0:
                xSouth = X+1
                ySouth = N-X+2*Y-1
                if 1 <= N-(ySouth-1) and N-(ySouth-1) <= N and 1 <= xSouth and xSouth <= N:
                    W[end-ySouth,xSouth-1] = 0
                xWest = X+1; yWest = N-X+2*Y
                if 1 <= N-(yWest-1) and N-(yWest-1) <= N and 1 <= xWest and xWest <= N:
                    W[end-yWest,xWest-1] = 0
            elif X%2 == 1:
                xSouth = X; ySouth = N-X+2*Y
                if 1 <= N-(ySouth-1) and N-(ySouth-1) <= N and 1 <= xSouth and xSouth <= N:
                    W[end-ySouth,xSouth-1] = 0
                xWest = X; yWest = N-X+2*Y-1
                if 1 <= N-(yWest-1) and N-(yWest-1) <= N and 1 <= xWest and xWest <= N:
                    W[end-yWest,xWest-1] = 0

    C = reduce_weight(W)
    C = numba.typed.List(C)
    M = shuffling(C).astype(int)
    fig = draw_dominos.draw_dominos(M,gap,str(edge),paths,dots,rotated,coloring,show_gap,dpi,show_figure)
    return fig,M,C
```

# Height map

```python
def height_function(M, a, b, c):
    """
    Compute the height function for the hexagon.

    Parameters:
        M : numpy.ndarray
            Matrix of size 2*(a+b+c-1) x 2*(a+b+c-1)
        a, b, c : int

    Returns:
        H : numpy.ndarray
            Matrix of size (c, b) with entries in {0, ..., a}
    """
    S = M.shape
    a,b,c = int(a),int(b),int(c)
    # assert S[0] == 2 * (a + b + c - 1) and S[0] == S[1], \
    #     "M must be of size 2(a+b+c-1) x 2(a+b+c-1)"

    H = np.zeros((c, b), dtype=int)

    for i in range(1, c + b):  # MATLAB: 1:(c+b-1)
        m1 = max(0, i - b)
        m2 = max(0, i - c)

        # MATLAB: M(end:-1:1, 2*i)
        col_2i = M[::-1, 2*i - 1]

        # MATLAB: find(...) → indices of nonzero elements (1-based)
        hvec = np.where(col_2i != 0)[0] + 1

        for j in range(1 + m1, len(hvec) - m2 + 1):
            # MATLAB indexing adjustments
            row_idx = c - j
            col_idx = i - j

            # MATLAB: M(end:-1:(end-hvec(j)+1), 2*i-1)
            h = hvec[j - 1]
            col_2i_minus_1 = M[::-1, 2*i - 2]
            segment = col_2i_minus_1[:h]

            H[row_idx, col_idx] = np.sum(segment)

    return H


def draw_hexagon2(n,w,a=1,b=1,c=1,skewed_grid=False,edge=0,paths=False,dots=False,coloring='standard',dpi=200,show_figure=True):
    w = w.astype(float)
    edge = str(edge)
    W = weight_hexagon.weight_hexagon(n,w,a,b,c)
    C = reduce_weight(W)
    M = shuffling(C).astype(int)
    fig = draw_lozenges.draw_lozenges(n,M,False,a,b,c,skewed_grid,edge,paths,dots,coloring,False,dpi,show_figure)
    return fig,M
```

```python
from scipy import ndimage
from scipy.signal import find_peaks

def count_hori_lines(H_rot,cut=0.5,plot=0,sigma=1,cut0=2,clean=0):
    N = H_rot.shape[0]
    img = H_rot[:,N//2-N//8:N//2+N//8+1]
    I = img>3
    img[I]=3
    I = img<cut0
    img[I]=0

    M = img.shape[1]

    img = ndimage.gaussian_filter(img, sigma=sigma)




    gray = img.astype(float)
    gray = (gray - gray.min()) / (gray.max() - gray.min())

    Peaks = []
    P_count = []
    for c in range(M):
        y = gray[:,c]
        y = y/max(y)
        y[y<cut]=0
        Peaks+= [find_peaks(y)[0]]
        P_count+=[len(Peaks[-1])]

    if clean:
        P = np.zeros(gray.shape)
        for k in range(M):
            P[:,k][Peaks[k]]=1

        ws = clean
        for m in range(P.shape[0]):
            for n in range(P.shape[1]):
                if P[m,n]:
                    A = P[m-ws:m+ws+1,n-ws:n+ws+1]
                    if np.sum(A)<(A.size)**0.5/2:
                        P[m,n]=0

        Peaks = [np.where(p==1)[0] for p in P.T]



    if plot>1:
        for k in range(gray.shape[1]):
            p = Peaks[k]
            n = len(p)
            plt.scatter(n*[k],p,c='r',s=1)
        # plt.imshow(P)
        # plt.show()
    #plt.scatter()
    if plot:
        plt.imshow(gray)
        plt.show()
    return round(np.median([len(p) for p in Peaks]))

count_hori_lines(H_rot,plot=2,clean=1)
```

```python
def comp_prob_fast(C):
    N = C.shape[0]//2
    Z = np.zeros((N,N))
    for m in range(N):
        for n in range(N):
            Z[m,n] = C[2*m,2*n]*C[2*m+1,2*n+1]/(C[2*m,2*n]*C[2*m+1,2*n+1]+C[2*m,2*n+1]*C[2*m+1,2*n])
    return Z

def comp_prob(C):
    N = C.shape[0]//2
    Z = np.zeros((N,N))
    for m in range(N):
        for n in range(N):
            w1,w2,w3,w4 = C[2*m,2*n],C[2*m,2*n+1],C[2*m+1,2*n],C[2*m+1,2*n+1]
            I14 = w1.imag + w4.imag
            I23 = w2.imag + w3.imag  
            if I14 > I23:
                Z[m,n] = 0
            elif I14 < I23:
                Z[m,n] = 1
            else:
                Z[m,n] = w1.real*w4.real/(w1.real*w4.real + w2.real*w3.real)
    return Z
    
```
