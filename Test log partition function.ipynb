```python
import RandomTilings as RT
import numpy as np
```

```python
RT.config.log_variant = True
```

```python
gap = [[4,-2,-1],[8,0,2],[7,0,2],[13,0,4],[2,-2,-2]] #[[9,1,5]]
A = RT.Aztec(7,gap=gap)
logg = A.log_partition_function()
print(logg) # 10.504519444188261+0j
print(np.round(np.exp(logg).real))
A.close()
```

```python
for i in range(1,10):
    A = RT.Aztec(i)
    logg = A.log_partition_function()
    print(np.round(np.exp(logg).real), logg)
    A.close()
    print(2**(i*(i+1)/2))
    print()
```
