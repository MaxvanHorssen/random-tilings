# Crystallization of the Aztec diamond

```python
import RandomTilings as RT
from numpy import pi, abs, sin
```

```python
def diff(theta,phi):
    return abs(sin((theta-phi)/2))

alpha = 0.
beta = pi/25
gamma = 4*pi/5
delta = 6*pi/5
epsilon = 8*pi/5

w = [[1+1j,diff(epsilon,gamma),diff(gamma,beta),diff(delta,gamma),diff(gamma,beta),diff(beta,gamma)],
[diff(gamma,beta),diff(beta,epsilon),1+4*1j,diff(beta,delta),1+1j,1+4*1j],
[diff(delta,gamma),diff(epsilon,delta),diff(delta,beta),1+1j,diff(delta,alpha),diff(alpha,delta)],
[diff(gamma,alpha),diff(alpha,epsilon),diff(beta,alpha),diff(alpha,delta),1+4*1j,1+1j]]

A = RT.Aztec(350,w)
A.shuffle()
A.plot(color_scheme='tropical',dpi=200,orientation='square')

ax = A.fig.axes[0]
ax.set_aspect(3/2)
ax.invert_xaxis()
ax.invert_yaxis()
```
