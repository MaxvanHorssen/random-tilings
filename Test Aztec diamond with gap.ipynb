```python
import RandomTilings as RT
import numpy as np
```

```python
RT.config.log_variant = True
```

```python
n = 100
m = int(10/300*n)
k = m
gap = [[2*n-10*m,k,n]]
print(gap)
A = RT.Aztec(n,gap=gap)
A.shuffle()
A.plot(path_width=0,gap_width=5,edge_width=0,orientation='diamond')
```
