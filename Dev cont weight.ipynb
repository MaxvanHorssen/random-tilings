```python
import numpy as np
import RandomTilings as RT
import matplotlib.pyplot as plt
```

```python
def const_weight(W_fun,N):
    W = np.zeros((2*N,2*N))
    for m in range(N):
        for n in range(N):
            W[2*m,2*n],W[2*m,2*n+1],W[2*m+1,2*n],W[2*m+1,2*n+1]=W_fun((n+1)/(N+1),(m+1)/(N+1))
    return W

```
